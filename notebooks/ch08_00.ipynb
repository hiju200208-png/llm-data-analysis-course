```python
from pathlib import Path
import sys

import pandas as pd
import numpy as np

# 프로젝트 루트와 notebooks 폴더 어느 곳에서 실행해도 경로를 찾습니다.
project_root = next(
    (path for path in (Path.cwd(), *Path.cwd().parents)
     if (path / "src" / "processing.py").is_file()
     and (path / "data" / "raw").is_dir()),
    None,
)
if project_root is None:
    raise FileNotFoundError("프로젝트 폴더 또는 notebooks 폴더에서 실행해 주세요.")
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

data_dir = project_root / "data" / "raw"
customers = pd.read_csv(data_dir / "customers.csv")
orders = pd.read_csv(data_dir / "orders.csv")
order_items = pd.read_csv(data_dir / "order_items.csv")
products = pd.read_csv(data_dir / "products.csv")

print(customers.head())
print(orders.head())
print(order_items.head())
print(products.head())
```

```python
#원본 구조 Evidence를 만든다.
raw_data= {"customers": customers, "order_items": order_items ,"products": products,"orders": orders}
print(raw_data)
```

```python
summary_list = []
for name, frame in raw_data.items():
    info = {
        "dataset": name,
        "rows": len(frame),
        "clums":frame.shape[1],
        "missing_valus":frame.isna().sum(),
        "duplicated_rows":frame.duplicated().sum(),
    }
    summary_list.append(info)
```

```python
# 각 데이터셋에서 기본 키로 사용할 컬럼
key_map = {
    "customers": "customer_id",
    "products": "product_id",
    "orders": "order_id",
    "order_items": "order_item_id",
}

key_checks = []

for dataset, key in key_map.items():
    frame = processed_data[dataset]
    missing_count = int(frame[key].isna().sum())
    duplicated_count = int(frame[key].duplicated().sum())

    key_checks.append({
        "dataset": dataset,
        "key": key,
        "missing_count": missing_count,
        "duplicated_count": duplicated_count,
    })

key_checks_df = pd.DataFrame(key_checks)
display(key_checks_df)
```

```python
# 주문 상세와 주문 정보를 결합하고 주문 항목별 매출액을 추가합니다.
order_sales = processed_data["order_items"].merge(
    processed_data["orders"][[
        "order_id", "customer_id", "order_date", "order_status"
    ]],
    on="order_id",
    how="left",
    validate="many_to_one",
)

# 전처리 결과에 line_total이 없을 때도 계산되도록 보장합니다.
order_sales["line_total"] = (
    order_sales["quantity"] * order_sales["unit_price"]
)

display(order_sales.head())
```
