### 실습 3. 머신러닝은 왜 텍스트를 숫자로 바꿀까?

머신러닝 알고리즘은 **숫자를 기반으로 계산**하기 때문에  
문자열 형태의 텍스트를 그대로 학습할 수 없습니다.

예를 들어,

`"데이터 분석을 위한 파이썬"`

이라는 문장을 머신러닝이 처리할 수 있도록 숫자로 변환해야 합니다.

#### 텍스트 벡터화 과정

도서 제목  
↓  
사용된 단어 확인  
↓  
단어마다 열(column) 생성  
↓  
단어의 등장 여부 또는 등장 횟수를 숫자로 기록  
↓  
숫자 벡터 생성

이처럼 **텍스트를 머신러닝이 계산할 수 있는 숫자 형태로 변환하는 과정**을  
**텍스트 벡터화(Vectorization)**라고 합니다.

> 핵심: 머신러닝은 텍스트 자체가 아니라 **텍스트를 숫자로 변환한 결과**를 이용해 학습합니다.

### 실습 4. 작은 예제로 텍스트 벡터화 이해하기

텍스트 벡터화의 원리를 쉽게 이해하기 위해 **3개의 문장**만 사용합니다.

#### 예제 문장

- 파이썬 데이터 분석
- 파이썬 머신러닝
- 데이터 분석 입문

#### 사용된 단어

`[데이터, 머신러닝, 분석, 입문, 파이썬]`

각 단어를 **열(column)**로 만들고, 문장에 해당 단어가 몇 번 등장했는지를 숫자로 표현합니다.

예를 들어,

**"파이썬 데이터 분석"**

→ `[1, 0, 1, 0, 1]`

각 숫자는 단어의 의미나 중요도가 아니라 **단어가 등장한 횟수**를 나타냅니다.

> **핵심:** 텍스트를 단어별 등장 횟수로 바꾸면 머신러닝이 처리할 수 있는 숫자 벡터가 됩니다.

### 실습 5. Bag of Words 이해하기

**Bag of Words(BoW)**는 문장에서 **단어의 순서보다 어떤 단어가 몇 번 등장했는지**에 초점을 맞추는 텍스트 표현 방식입니다.

#### 예시

- 파이썬 데이터 분석
- 데이터 파이썬 분석

두 문장은 단어의 **순서가 다르지만**, 사용된 단어와 등장 횟수가 같기 때문에 BoW에서는 **동일한 Count 벡터**로 표현될 수 있습니다.

#### Bag of Words 특징

- 단어별 **등장 횟수**를 숫자로 표현
- 구조가 단순하고 이해하기 쉬움
- 단어 순서는 고려하지 않음
- 문맥 정보를 충분히 표현하기 어려움

> **핵심:** Bag of Words는 **단어의 순서보다는 어떤 단어가 몇 번 등장했는지**를 중요하게 보는 방식입니다.
