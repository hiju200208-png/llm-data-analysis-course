### 실습 3. 머신러닝은 왜 텍스트를 숫자로 바꿀까?

머신러닝 알고리즘은 **숫자를 기반으로 계산**하기 때문에  
문자열 형태의 텍스트를 그대로 학습할 수 없습니다.

예를 들어,

`"데이터 분석을 위한 파이썬"`

이라는 문장을 머신러닝이 처리할 수 있도록 숫자로 변환해야 합니다.

#### 텍스트 벡터화 과정

도서 제목  
↓  
사용된 단어 확인  
↓  
단어마다 열(column) 생성  
↓  
단어의 등장 여부 또는 등장 횟수를 숫자로 기록  
↓  
숫자 벡터 생성

이처럼 **텍스트를 머신러닝이 계산할 수 있는 숫자 형태로 변환하는 과정**을  
**텍스트 벡터화(Vectorization)**라고 합니다.

> 핵심: 머신러닝은 텍스트 자체가 아니라 **텍스트를 숫자로 변환한 결과**를 이용해 학습합니다.

### 실습 4. 작은 예제로 텍스트 벡터화 이해하기

텍스트 벡터화의 원리를 쉽게 이해하기 위해 **3개의 문장**만 사용합니다.

#### 예제 문장

- 파이썬 데이터 분석
- 파이썬 머신러닝
- 데이터 분석 입문

#### 사용된 단어

`[데이터, 머신러닝, 분석, 입문, 파이썬]`

각 단어를 **열(column)**로 만들고, 문장에 해당 단어가 몇 번 등장했는지를 숫자로 표현합니다.

예를 들어,

**"파이썬 데이터 분석"**

→ `[1, 0, 1, 0, 1]`

각 숫자는 단어의 의미나 중요도가 아니라 **단어가 등장한 횟수**를 나타냅니다.

> **핵심:** 텍스트를 단어별 등장 횟수로 바꾸면 머신러닝이 처리할 수 있는 숫자 벡터가 됩니다.

### 실습 5. Bag of Words 이해하기

**Bag of Words(BoW)**는 문장에서 **단어의 순서보다 어떤 단어가 몇 번 등장했는지**에 초점을 맞추는 텍스트 표현 방식입니다.

#### 예시

- 파이썬 데이터 분석
- 데이터 파이썬 분석

두 문장은 단어의 **순서가 다르지만**, 사용된 단어와 등장 횟수가 같기 때문에 BoW에서는 **동일한 Count 벡터**로 표현될 수 있습니다.

#### Bag of Words 특징

- 단어별 **등장 횟수**를 숫자로 표현
- 구조가 단순하고 이해하기 쉬움
- 단어 순서는 고려하지 않음
- 문맥 정보를 충분히 표현하기 어려움

> **핵심:** Bag of Words는 **단어의 순서보다는 어떤 단어가 몇 번 등장했는지**를 중요하게 보는 방식입니다.

### 실습 7. 작은 예제에 CountVectorizer 적용하기

```python
from sklearn.feature_extraction.text import CountVectorizer

# 작은 예제 문서
sample_docs = [
    "파이썬 데이터 분석",
    "파이썬 머신러닝",
    "데이터 분석 입문",
]

# CountVectorizer 생성
count_vectorizer = CountVectorizer()

# fit + transform
X_count_sample = count_vectorizer.fit_transform(sample_docs)

# 결과 확인
print(X_count_sample)
```

### 실습 8. 생성된 단어 사전 확인하기

```python
feature_names = count_vectorizer.get_feature_names_out()

print(feature_names)
```

### 실습 9. 단어-문서 행렬 확인하기

```python
import pandas as pd

# 숫자 배열 확인
print(X_count_sample.toarray())

# DataFrame으로 변환
sample_count_df = pd.DataFrame(
    X_count_sample.toarray(),
    columns=feature_names,
    index=sample_docs,
)

sample_count_df
```

### 실습 10. 같은 단어가 여러 번 나오면 어떻게 될까?

```python
repeat_docs = [
    "파이썬 파이썬 데이터",
    "데이터 분석",
]

repeat_vectorizer = CountVectorizer()

X_repeat = repeat_vectorizer.fit_transform(repeat_docs)

repeat_df = pd.DataFrame(
    X_repeat.toarray(),
    columns=repeat_vectorizer.get_feature_names_out(),
    index=repeat_docs,
)

repeat_df
```

### 실습 12. 실제 도서 제목에 CountVectorizer 적용하기

```python
from sklearn.feature_extraction.text import CountVectorizer

# CountVectorizer 생성
count_vectorizer = CountVectorizer()

# 실제 도서 제목을 숫자 벡터로 변환
X_count = count_vectorizer.fit_transform(titles)

# 생성된 단어 목록
count_terms = count_vectorizer.get_feature_names_out()

# 크기 확인
print("문서 수:", X_count.shape[0])
print("단어 수:", X_count.shape[1])
print("행렬 크기:", X_count.shape)
```

### 실습 13. vocabulary_ 확인하기

Vectorizer 내부에는 단어와 열 번호의 대응 정보가 있습니다.

```python
vocabulary = list(count_vectorizer.vocabulary_.items())[:20]

for word, column_index in vocabulary:
    print(word, "→", column_index)
```

### 실습 14. 희소 행렬(Sparse Matrix) 이해하기

```python
# X_count의 데이터 타입 확인
print(type(X_count))

# 희소 행렬 확인
print("타입:", type(X_count))
print("행렬 크기:", X_count.shape)
print("0이 아닌 값의 개수:", X_count.nnz)

# 처음 5개 문서만 확인
print(X_count[:5].toarray())
```

### 실습 15. 실제 데이터의 첫 번째 제목 벡터 확인하기

```python
# 첫 번째 도서 제목 확인
print("첫 번째 도서 제목:")
print(titles.iloc[0])

# 첫 번째 제목의 벡터 가져오기
first_row = X_count.getrow(0)

# 0보다 큰 값이 있는 열 번호와 등장 횟수
indices = first_row.indices
values = first_row.data

# 단어와 등장 횟수 연결
first_title_terms = [
    (count_terms[index], value)
    for index, value in zip(indices, values)
]

print("\n벡터화 결과:")
print(first_title_terms)
```

### 실습 16. 전체 데이터에서 많이 등장한 단어 확인하기

```python
import numpy as np
import pandas as pd

# 각 단어의 전체 등장 횟수 계산
count_sums = np.asarray(X_count.sum(axis=0)).ravel()

# DataFrame 생성
count_summary = pd.DataFrame({
    "단어": count_terms,
    "전체등장횟수": count_sums,
})

# 등장 횟수가 많은 순서대로 정렬
count_summary = count_summary.sort_values(
    "전체등장횟수",
    ascending=False,
).reset_index(drop=True)

# 상위 30개 확인
count_summary.head(30)
```

### 실습 17. Count 상위 단어 저장하기

```python
count_top30 = count_summary.head(30)

count_top30.to_csv(

    "chapter03_count_top_terms.csv",

    index=False,

    encoding="utf-8-sig",

)

pd.read_csv(

    "chapter03_count_top_terms.csv",

    encoding="utf-8-sig",

).head()
```

### 실습 23. 작은 예제로 TfidfVectorizer 적용하기

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# TF-IDF Vectorizer 생성
tfidf_sample_vectorizer = TfidfVectorizer()

# 텍스트를 TF-IDF 값으로 변환
X_tfidf_sample = tfidf_sample_vectorizer.fit_transform(sample_docs)

# 단어 목록 확인
tfidf_sample_terms = tfidf_sample_vectorizer.get_feature_names_out()

# DataFrame으로 변환
tfidf_sample_df = pd.DataFrame(
    X_tfidf_sample.toarray(),
    columns=tfidf_sample_terms,
    index=sample_docs,
)

# 소수점 셋째 자리까지 확인
tfidf_sample_df.round(3)
```

### 실습 24. TfidfVectorizer가 만든 단어 확인하기

```python
# TF-IDF가 만든 단어 목록 확인
print(tfidf_sample_vectorizer.get_feature_names_out())

tfidf_terms = tfidf_sample_vectorizer.get_feature_names_out()

for index, word in enumerate(tfidf_terms):
    print(index, "→", word)
```
